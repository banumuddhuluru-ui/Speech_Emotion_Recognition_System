```python
# ============================================================
# SPEECH EMOTION RECOGNITION SYSTEM
# Audio Analysis + Gradio
# ============================================================

!pip install -q gradio librosa numpy

import gradio as gr
import librosa
import numpy as np


# ------------------------------------------------------------
# EMOTION ANALYSIS
# ------------------------------------------------------------

def recognize_emotion(audio_file):

    if audio_file is None:
        return (
            "Please upload or record an audio file.",
            "0%",
            "Waiting for audio input."
        )

    try:
        # Load audio
        y, sr = librosa.load(audio_file, sr=None)

        # Extract audio features
        rms = np.mean(librosa.feature.rms(y=y))
        zcr = np.mean(librosa.feature.zero_crossing_rate(y))

        # Simple demo-based classification
        if rms > 0.08 and zcr > 0.08:
            emotion = "Excited / Happy"
            confidence = 75

        elif rms < 0.03:
            emotion = "Calm / Sad"
            confidence = 70

        elif zcr > 0.06:
            emotion = "Energetic"
            confidence = 72

        else:
            emotion = "Neutral"
            confidence = 68

        analysis = (
            f"Audio energy (RMS): {rms:.4f}\n"
            f"Speech variation (ZCR): {zcr:.4f}\n"
            f"The system analyzed acoustic features to estimate "
            f"the emotional state."
        )

        return (
            emotion,
            f"{confidence}%",
            analysis
        )

    except Exception as e:

        return (
            "Unable to analyze audio",
            "0%",
            str(e)
        )


# ------------------------------------------------------------
# CUSTOM CSS
# ------------------------------------------------------------

css = """
body {
    background-color: #0b0f19 !important;
}

.gradio-container {
    background-color: #0b0f19 !important;
    color: white !important;
    max-width: 1200px !important;
}

h1 {
    color: white !important;
    font-size: 34px !important;
}

h2, h3, p, label {
    color: white !important;
}

textarea, input {
    background-color: #374151 !important;
    color: white !important;
}

button {
    background-color: #5146e5 !important;
    color: white !important;
    font-weight: bold !important;
}
"""


# ------------------------------------------------------------
# WEB APPLICATION
# ------------------------------------------------------------

with gr.Blocks(
    theme=gr.themes.Base(),
    css=css,
    title="Speech Emotion Recognition System"
) as demo:

    gr.Markdown(
        """
        # 🎭 Speech Emotion Recognition System

        ### Analyze speech and estimate the speaker's emotional state

        This application analyzes acoustic features of speech
        and provides an estimated emotion.
        """
    )

    gr.Markdown(
        """
        **Features:**

        - 🎤 Audio input
        - 📊 Acoustic feature analysis
        - 🎭 Emotion estimation
        - 🧠 Speech analysis
        """
    )

    with gr.Row():

        # INPUT
        with gr.Column():

            gr.Markdown("### 🎙️ Input Speech")

            audio_input = gr.Audio(
                sources=["microphone", "upload"],
                type="filepath",
                label="Record or Upload Audio"
            )

            analyze_button = gr.Button(
                "🎭 Analyze Emotion",
                variant="primary"
            )

        # OUTPUT
        with gr.Column():

            gr.Markdown("### 🔎 Emotion Result")

            emotion_output = gr.Textbox(
                label="Detected Emotion",
                interactive=False
            )

            confidence_output = gr.Textbox(
                label="Confidence",
                interactive=False
            )

            analysis_output = gr.Textbox(
                label="Audio Analysis",
                lines=5,
                interactive=False
            )

    # Button action
    analyze_button.click(
        fn=recognize_emotion,
        inputs=audio_input,
        outputs=[
            emotion_output,
            confidence_output,
            analysis_output
        ]
    )


# ------------------------------------------------------------
# LAUNCH
# ------------------------------------------------------------

demo.launch(share=True)
```
